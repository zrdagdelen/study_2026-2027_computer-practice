```julia
#Факторизации (LU, QR, Собственные)
# Решение СЛАУ
A = rand(3, 3); x = fill(1.0, 3); b = A * x
println("Решение A \\ b: ", A \ b)

# LU-разложение
Alu = lu(A)
println("L-фактор:\n", Alu.L)
println("U-фактор:\n", Alu.U)

# QR-разложение
Aqr = qr(A)
println("Матрица Q ортогональна? ", Aqr.Q' * Aqr.Q ≈ Matrix(I, 3, 3))

# Симметризация и спектральное разложение
Asym = A + A'
eig = eigen(Asym)
println("Собственные значения: ", eig.values)
println("Проверка: inv(V)*A*V =\n", inv(eig.vectors) * Asym * eig.vectors)
```

```julia
# BenchmarkTools (спец. структуры)
n = 1000
A = randn(n,n)
Asym = A + A'
Asym_noisy = copy(Asym)
Asym_noisy[1,2] += 5eps()
Asym_explicit = Symmetric(Asym_noisy)

# Сравнение скорости
println("Обычная симметричная:")
@btime eigvals(Asym);
println("Зашумленная:")
@btime eigvals(Asym_noisy);
println("Явно указанная Symmetric:")
@btime eigvals(Asym_explicit);
```

```julia
println("--- С двумя неизвестными ---")
# a) x+y=2, x-y=3
println("a) ", [1 1; 1 -1] \ [2, 3])
# b) x+y=2, 2x+2y=4 (Бесконечно много решений. Ранг = 1)
println("b) Ранг: ", rank([1 1; 2 2]), " (решений бесконечно много)")
# c) x+y=2, 2x+2y=5 (Нет решений)
println("c) Ранг A: ", rank([1 1; 2 2]), ", Ранг A|b: ", rank([1 1 2; 2 2 5]), " (решений нет)")
# d) x+y=1, 2x+2y=2, 3x+3y=3 (Псевдорешение)
println("d) ", pinv([1 1; 2 2; 3 3]) * [1, 2, 3])
# e) x+y=2, 2x+y=1, x-y=3
println("e) ", pinv([1 1; 2 1; 1 -1]) * [2, 1, 3])
# f) x+y=2, 2x+y=1, 3x+2y=3
println("f) ", pinv([1 1; 2 1; 3 2]) * [2, 1, 3])
```

```julia
# 3. Собственные значения матрицы А и оценка эффективности
# Задаем матрицу А
A = [
    140  97  74 168 131;
     97 106  89 131  36;
     74  89 152 144  71;
    168 131 144  54 142;
    131  36  71 142  36
]

# Находим собственные значения
vals = eigvals(A)
println("Собственные значения матрицы А:\n", vals)
```

```julia
# Создаем диагональную матрицу из собственных значений
D = Diagonal(vals)
println("\nДиагональная матрица (из собственных значений А):\n", D)

# Создаем нижнедиагональную (нижнетреугольную) матрицу из матрицы А
# (по сути, оставляем только элементы на главной диагонали и ниже)
L = LowerTriangular(A)
println("\nНижнетреугольная матрица из А:\n", L)
```

4.4.4. Модель Леонтьева (продуктивность матрицы)

```julia
# 1. Проверка по определению
println("--- 1. Проверка по определению ---")
# По определению: берем неотрицательный вектор y и проверяем, что решение x >= 0.
# Возьмем y = [1, 1]. Если x получится с минусом, то матрица не продуктивна.

y_test = [1.0, 1.0]

# a) A = [1 2; 3 4]
A1 = [1.0 2.0; 3.0 4.0]
E1 = Matrix{Float64}(I, 2, 2)
x1 = (E1 - A1) \ y_test
println("a) Решение x = ", x1, " -> продуктивна? ", all(x1 .>= 0))
```

```julia
# b) A = 1/2 * [1 2; 3 4]
A2 = (1/2) * [1.0 2.0; 3.0 4.0]
E2 = Matrix{Float64}(I, 2, 2)
x2 = (E2 - A2) \ y_test
println("b) Решение x = ", x2, " -> продуктивна? ", all(x2 .>= 0))
```

```julia
# c) A = 1/10 * [1 2; 3 4]
A3 = (1/10) * [1.0 2.0; 3.0 4.0]
E3 = Matrix{Float64}(I, 2, 2)
x3 = (E3 - A3) \ y_test
println("c) Решение x = ", x3, " -> продуктивна? ", all(x3 .>= 0))
```

```julia
# Функция для проверки критериев
function check_criteria(A, name)
    n = size(A, 1)
    E = Matrix{Float64}(I, n, n)

    # Проверка 2-го критерия: все элементы (E-A)^-1 >= 0
    B = inv(E - A)
    crit2 = all(B .>= 0)

    # Проверка 3-го критерия (спектральный): max|λ| < 1
    lambdas = eigvals(A)
    max_lambda = maximum(abs.(lambdas))
    crit3 = max_lambda < 1

    println("Матрица $name:")
    println("  Собственные значения: ", lambdas)
    println("  Критерий 2 (матрица (E-A)⁻¹ >= 0): ", crit2)
    println("  Критерий 3 (спектральный max|λ| < 1): ", crit3, " (max|λ| = ", round(max_lambda, digits=3), ")")
    println()
end
```

```julia


#2 и 3. Проверка по критериям для матриц [1 2; 3 1] и их вариаций
println("--- 2 и 3. Проверка матриц из пунктов 2 и 3 ---")

# Задаем базовую матрицу для пунктов 2 и 3
M_base = [1.0 2.0; 3.0 1.0]

# a)
A_2a = M_base
check_criteria(A_2a, "2a/3a")

# b)
A_2b = (1/2) * M_base
check_criteria(A_2b, "2b/3b")

# c)
A_2c = (1/10) * M_base
check_criteria(A_2c, "2c/3c")

# 3d. Проверка матрицы 3x3
println("--- 3d. Матрица 3x3 ---")
A_3d = [
    0.1 0.2 0.3;
    0.0 0.1 0.2;
    0.0 0.1 0.3
]

check_criteria(A_3d, "3d")
```
