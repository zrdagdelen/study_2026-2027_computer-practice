# Раздел 2.2.1 — Кортежи

```julia
# пустой кортеж
println(())
println("---------")

# кортеж из строк
favoritelang = ("Python", "Julia", "R")
println(favoritelang)
println("---------")

# кортеж из целых чисел
x1 = (1, 2, 3)
println(x1)
println("---------")

# кортеж из разных типов
x2 = (1, 2.0, "tmp")
println(x2)
println("---------")

# именованный кортеж
x3 = (a=2, b=1+2)
println(x3)
println("---------")

# длина кортежа x2
println(length(x2))
println("---------")

# обращение к элементам x2
println(x2[1], x2[2], x2[3])
println("---------")

# операция с элементами x1
c = x1[2] + x1[3]
println(c)
println("---------")

# обращение к элементам именованного кортежа x3
println(x3.a, x3.b, x3[2])
println("---------")

# проверка вхождения
in("tmp", x2), 0 in x2
```

# Задание 2.4, пункт 1 — Множества

```julia
# Даны множества A, B, C
A = Set([0, 3, 4, 9])       # создаём множество A из перечисленных элементов
B = Set([1, 3, 4, 7])       # создаём множество B
C = Set([0, 1, 2, 4, 7, 8, 9])  # создаём множество C

# P = A∩B ∪ A∩C ∪ B∩C
# intersect(a,b) — пересечение множеств
# union(a,b,c)   — объединение нескольких множеств
P = intersect(union(A, B), union(A, C), union(B, C))
# ↑ сначала объединяем A∪B, A∪C, B∪C, потом пересекаем эти три результата
println("P = ", P)          # выводим результат P на экран
```

# Задание 2.4, пункт 3.14 — большая задача

```julia
# n = 250; x и y — случайные целые из [0,999]
n = 250
x = rand(0:999, n)              # вектор из n случайных целых от 0 до 999
y = rand(0:999, n)              # аналогично для y

# (y2-y1, ..., yn-y_{n-1})
v1 = [y[i+1] - y[i] for i in 1:n-1]
# ↑ для i от 1 до n-1: разность соседних элементов y
println("v1 (первые 5): ", v1[1:5])

# (x1+2x2-x3, x2+2x3-x4, ..., x_{n-2}+2x_{n-1}-x_n)
v2 = [x[i] + 2*x[i+1] - x[i+2] for i in 1:n-2]
# ↑ для i от 1 до n-2: комбинация трёх подряд идущих элементов x
println("v2 (первые 5): ", v2[1:5])

# (sin(y1)/cos(x2), sin(y2)/cos(x3), ..., sin(y_{n-1})/cos(x_n))
v3 = [sin(y[i]) / cos(x[i+1]) for i in 1:n-1]
# ↑ для i от 1 до n-1: sin(y[i]) / cos(x[i+1])
println("v3 (первые 5): ", v3[1:5])

# сумма e^{-x_{i+1}} / (x_i + 10), i = 1..n-1
s = sum(exp(-x[i+1]) / (x[i] + 10) for i in 1:n-1)
# ↑ sum(...) складывает все значения выражения для i от 1 до n-1
println("Сумма: ", s)
```

# Задание 2.4, пункт 6.3

```julia
# 1 + 2/3 + (2*4)/(3*5) + (2*4*6)/(3*5*7) + ...
# каждый следующий член: числитель умножается на 2k, знаменатель — на (2k+1)
terms = [1.0]                   # список членов, начинаем с 1
numer = 1.0                     # текущий числитель
denom = 1.0                     # текущий знаменатель
for k in 2:20                   # считаем 20 членов ряда
    global numer *= 2k          # numer = numer * 2k (умножаем на следующее чётное)
    global denom *= (2k + 1)    # denom = denom * (2k+1) (умножаем на следующее нечётное)
    push!(terms, numer / denom) # добавляем новый член в список
end
println("6.3 (первые 5 членов): ", terms[1:5])
println("6.3 сумма: ", sum(terms))   # сумма всех членов
```
