### Задание 1. Функции ввода/вывода

```julia
# print и println
print("Hello ")
println("World!")

# show
show("Hello")

# readline (ввод с клавиатуры)
println("Введите ваше имя:")
name = readline()
println("Привет, $name !")

# запись в файл
open("test.txt", "w") do f
    write(f, "Это тестовая запись")
end

# чтение из файла
open("test.txt", "r") do f
    content = read(f, String)
    println("Содержимое файла: $content")
end

# readlines (чтение всех строк)
open("test.txt", "r") do f
    lines = readlines(f)
    println("Строки: $lines")
end
```

### Задание 2. Функция parse()

```julia
# преобразование строки в Int
x = parse(Int64, "456")
println(x, " тип: ", typeof(x))

# преобразование в Float
y = parse(Float64, "3.1415")
println(y, " тип: ", typeof(y))

# попытка преобразовать нечисловую строку (вызовет ошибку)
parse(Int64, "abc")
```

### Задание 3. Математические операции

```julia
a = 10
b = 3

println("a = $a, b = $b")
println("Сложение: ", a + b)
println("Вычитание: ", a - b)
println("Умножение: ", a * b)
println("Деление: ", a / b)
println("Целочисленное деление: ", div(a, b))
println("Остаток от деления: ", rem(a, b))
println("Возведение в степень: ", a^b)
println("Квадратный корень из a: ", sqrt(a))
println("a > b: ", a > b)
println("a == b: ", a == b)
println("Логическое И: ", (a > 5) && (b < 5))
println("Логическое ИЛИ: ", (a > 5) || (b > 5))
```

### Задание 4. Операции с матрицами и векторами

```julia
using LinearAlgebra
A = [1 2; 3 4]
B = [5 6; 7 8]
v = [1, 2]

println("Матрица A:")
display(A)
println("Матрица B:")
display(B)
println("Вектор v:")
display(v)

println("Сложение матриц A + B:")
display(A + B)

println("Вычитание матриц A - B:")
display(A - B)

println("Умножение матрицы на скаляр (2 * A):")
display(2 * A)

println("Транспонирование A':")
display(A')

println("Скалярное произведение v · v: ", dot(v,v))

println("Умножение матрицы на вектор A * v:")
display(A * v)
```
