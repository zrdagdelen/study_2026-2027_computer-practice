## 5.2.2. Опции при построении графика (sin(x) и ряд Тейлора)

```julia
pyplot()

sin_theor(x) = sin(x)
sin_taylor(x) = sum((-1)^i * x^(2i+1) / factorial(2i+1) for i in 0:4)

# два графика на одном рисунке
plot(sin_theor, -2π, 2π,
     label="sin(x), теоретическое значение",
     xlabel="x", ylabel="y",
     title="sin(x) и разложение в ряд Тейлора",
     line=(:black, 1.0, 2, :dash))

plot!(sin_taylor, -2π, 2π,
      label="sin(x), ряд Тейлора",
      line=(:blue, 2.0),
      fillrange=0, fillalpha=0.3, fillcolor=:lightgoldenrod,
      xrotation=rad2deg(pi/4),
      background_color=:ivory)
```

## 5.2.4. Аппроксимация данных

```julia
x = collect(0:0.01:9.99)
y = exp.(ones(1000) .+ x) .+ 4000 .* randn(1000)

scatter(x, y, markersize=3, alpha=0.8, legend=false,
        title="Исходные данные", xlabel="x", ylabel="y")
```

```julia
A = [ones(1000) x x.^2 x.^3 x.^4 x.^5]
c = A \ y
f_approx = c[1].*ones(1000) .+ c[2].*x .+ c[3].*x.^2 .+ c[4].*x.^3 .+ c[5].*x.^4 .+ c[6].*x.^5

scatter(x, y, markersize=2, alpha=0.4, label="данные")
plot!(x, f_approx, color=:red, linewidth=2,
      label="полином 5-й степени",
      title="Аппроксимация полиномом 5-й степени",
      xlabel="x", ylabel="y", legend=:topleft)
```

## 5.2.9. Линии уровня

$g(x,y) = (3x + y^2)|sin(x) + cos(y)|$

```julia
x = 1:0.5:20
y = 1:0.5:10
g(x, y) = (3x + y^2) * abs(sin(x) + cos(y))

plot(x, y, g, linetype=:surface,
     title="Поверхность g(x,y) = (3x + y²)|sin(x)+cos(y)|",
     xlabel="x", ylabel="y", zlabel="g(x,y)")
```

```julia
contour(x, y, g,
        title="Линии уровня g(x,y)",
        xlabel="x", ylabel="y")
```

```julia
p = contour(x, y, g, fill=true,
            title="Линии уровня с заливкой",
            xlabel="x", ylabel="y")
plot(p)
```

## 5.2.12. Errorbars


```julia
using Plots
gr()

sds = [1, 1/2, 1/4, 1/8, 1/16, 1/32]
n = 10
y = [mean(sd .* randn(n)) for sd in sds]
errs = 1.96 .* sds ./ sqrt(n)

plot(y, ylims=(-1, 1), err=errs,
     title="Значения с доверительными интервалами",
     xlabel="номер точки", ylabel="y",
     label="y ± 1.96σ", legend=:topright)
```

## 5.2.13. Гистограммы (Distributions)


```julia
ages = rand(15:55, 1000)
histogram(ages,
          title="Гистограмма случайных возрастов",
          xlabel="Возраст (лет)", ylabel="Количество",
          label="равномерное", legend=:topright)
```

```julia
d = Normal(35.0, 10.0)
ages = rand(d, 1000)
histogram(ages,
          label="Распределение по возрастам (года)",
          xlabel="Возраст (лет)", ylabel="Количество",
          title="Гистограмма нормального распределения", legend=:topright)
```

```julia
using Plots
gr()

using Distributions
d1 = Normal(10.0, 5.0); d2 = Normal(35.0, 10.0); d3 = Normal(60.0, 5.0)
N = 1000
ages = Float64[]
append!(ages, rand(d1, Int64(ceil(N/2))))
append!(ages, rand(d2, N))
append!(ages, rand(d3, Int64(ceil(N/3))))

histogram(ages, bins=50,
          label="Распределение по возрастам (года)",
          xlabel="Возраст (лет)", ylabel="Количество",
          title="Распределение по возрастам (года)", legend=:topright)
```

### Задание 4.
Вектор $x = (-2,-1,0,1,2), y = x^3−3x$ в 4-х панелях

```julia
using Plots
gr()

x4 = [-2, -1, 0, 1, 2]
y4 = x4.^3 .- 3 .* x4

# для плавной кривой берём много точек на том же отрезке
x_smooth = range(-2, 2, length=200)
y_smooth = x_smooth.^3 .- 3 .* x_smooth

q1 = scatter(x4, y4, title="Точки", legend=false)
q2 = plot(x4, y4, title="Линия", legend=false)
q3 = plot(x4, y4, seriestype=:path, marker=:circle, title="Линия + точки", legend=false)
q4 = plot(x_smooth, y_smooth, title="Кривая", legend=false)

plot(q1, q2, q3, q4, layout=(2,2), size=(900,600),
     plot_title="y = x³ − 3x")

savefig("figure_Dagdelen.png")
```

### Задание 5.
$x = (3, 3.1, ..., 6), y_1 = \pi x, y_2 = \exp ^x·cos(x)$

```julia
x5 = 3:0.1:6
y1 = π .* x5
y2 = exp.(x5) .* cos.(x5)

# Первый вариант: оба графика на одном рисунке
plot(x5, [y1 y2],
     label=["y1 = πx" "y2 = eˣ·cos(x)"],
     xlabel="x", ylabel="y",
     title="Два графика на одном рисунке",
     color=[:blue :red],
     linewidth=2,
     grid=true,
     size=(800,500))
```

```julia
# Второй вариант: две панели (так как масштабы y1 и y2 сильно различаются —
# недостаток первого построения: y1 почти не видна на фоне y2)
pa = plot(x5, y1, title="y1 = πx", xlabel="x", ylabel="y1", color=:blue, legend=false)
pb = plot(x5, y2, title="y2 = eˣ·cos(x)", xlabel="x", ylabel="y2", color=:red, legend=false)
plot(pa, pb, layout=(1,2), size=(900,400),
     plot_title="Два графика на отдельных панелях")
```

Недостаток первого построения: функции имеют разный порядок значений ($y_1$ ≈ 9–19, $y_2$ достигает ~200 и меняет знак), поэтому на общей оси ординат кривая y₁ = πx выглядит почти горизонтальной линией и плохо читается. Поэтому лучше строить в двух панелях (второй вариант) или на двух осях ординат.
